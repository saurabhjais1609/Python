```python
# Input and Output
# The print() function is used for output in various formats and the input() function enables interaction with users.
```

```python
# input() function is used to take user input. By default, it returns the user input in form of a string. 
name = input("Enter your name: ")
print("Hello,", name, "! Welcome!")
```

```python
# print() function allows us to display text, variables and expressions on the console.
print("Hello World!")
```

```python
# Variables
s = "Brad"
print(s)

s = "Anjelina"
age = 25
city = "New York"
print(s, age, city)
```

```python
# Taking multiple inputs at once
x, y = input("Enter two numbers: ").split()
print(x, y)
```

```python
# Taking different types of input
i = int(input("How old are you?:"))
f = float(input("Evaluate 7/2:"))
print("Your age:", i)
print("7/2:", f)
```
